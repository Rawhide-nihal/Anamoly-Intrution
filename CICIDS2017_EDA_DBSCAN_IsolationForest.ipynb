```python
# Clean column names

df.columns = df.columns.str.strip()

print("Column names cleaned successfully!")
print(df.columns.tolist())
```

```python
# Identify numerical features

numeric_columns = df.select_dtypes(include=np.number).columns

print("Number of numerical features:", len(numeric_columns))

print("\nNumerical features:")
for i, col in enumerate(numeric_columns, start=1):
    print(f"{i}. {col}")
```

```python
# Statistical summary of numerical features

feature_stats = df[numeric_columns].describe().T

feature_stats["range"] = feature_stats["max"] - feature_stats["min"]

feature_stats.sort_values("range", ascending=False).head(15)
```

```python
# Distribution of selected network features

selected_features = [
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Flow Bytes/s"
]

for feature in selected_features:
    if feature in df.columns:
        plt.figure(figsize=(8, 5))
        plt.hist(df[feature].replace([np.inf, -np.inf], np.nan).dropna(), bins=50)
        plt.title(f"Distribution of {feature}")
        plt.xlabel(feature)
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()
```

```python
# Correlation analysis

correlation_features = [
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Fwd Packet Length Mean",
    "Bwd Packet Length Mean"
]

correlation_features = [
    col for col in correlation_features if col in df.columns
]

correlation_matrix = df[correlation_features].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Between Network Traffic Features")
plt.tight_layout()
plt.show()
```

```python
# Outlier analysis using Flow Duration

feature = "Flow Duration"

if feature in df.columns:
    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[feature] < lower_bound) |
        (df[feature] > upper_bound)
    ]

    print("Feature:", feature)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Potential outliers:", len(outliers))
```

```python
# Load a sample from all CIC-IDS2017 files

sample_size = 5000
dataframes = []

for file in csv_files:
    print(f"Loading: {file.name}")

    temp = pd.read_csv(
        file,
        nrows=sample_size,
        low_memory=False
    )

    temp.columns = temp.columns.str.strip()
    dataframes.append(temp)

combined_df = pd.concat(dataframes, ignore_index=True)

print("\nCombined dataset created!")
print("Rows:", combined_df.shape[0])
print("Columns:", combined_df.shape[1])
```

```python
# Analyze traffic labels in the combined dataset

label_col = [col for col in combined_df.columns if "label" in col.lower()][0]

print("Label column:", repr(label_col))
print("\nTraffic distribution:")
print(combined_df[label_col].value_counts())
```

```python
# Visualize traffic distribution

plt.figure(figsize=(12, 6))

combined_df[label_col].value_counts().plot(kind="bar")

plt.title("CIC-IDS2017 Traffic Distribution")
plt.xlabel("Traffic Type")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Check combined dataset quality

print("Missing values:", combined_df.isnull().sum().sum())
print("Duplicate rows:", combined_df.duplicated().sum())

numeric_combined = combined_df.select_dtypes(include=np.number)

print(
    "Infinite values:",
    np.isinf(numeric_combined).sum().sum()
)
```

```python
# Check feature data types

print("Numerical features:", len(combined_df.select_dtypes(include=np.number).columns))
print("Non-numerical features:", len(combined_df.select_dtypes(exclude=np.number).columns))

print("\nNon-numerical columns:")
print(combined_df.select_dtypes(exclude=np.number).columns.tolist())
```

```python
# Check for columns with zero variance

numeric_features = combined_df.select_dtypes(include=np.number)

zero_variance = numeric_features.columns[
    numeric_features.nunique() <= 1
]

print("Zero-variance features:", len(zero_variance))

if len(zero_variance) > 0:
    print(zero_variance.tolist())
```

```python
# Remove zero-variance features

zero_variance_features = [
    'Bwd PSH Flags',
    'Fwd URG Flags',
    'Bwd URG Flags',
    'CWE Flag Count',
    'Fwd Avg Bytes/Bulk',
    'Fwd Avg Packets/Bulk',
    'Fwd Avg Bulk Rate',
    'Bwd Avg Bytes/Bulk',
    'Bwd Avg Packets/Bulk',
    'Bwd Avg Bulk Rate'
]

combined_df = combined_df.drop(columns=zero_variance_features)

print("Removed features:", len(zero_variance_features))
print("Remaining columns:", combined_df.shape[1])
```

```python
# Separate features and labels

X = combined_df.drop(columns=[label_col])
y = combined_df[label_col]

print("Feature matrix shape:", X.shape)
print("Label vector shape:", y.shape)

print("\nLabels:")
print(y.value_counts())
```

```python
# Replace infinite values with NaN

X = X.replace([np.inf, -np.inf], np.nan)

print("Missing values before cleaning:", X.isnull().sum().sum())
```

```python
# Fill missing values using median

X = X.fillna(X.median(numeric_only=True))

print("Missing values after cleaning:", X.isnull().sum().sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Scaled feature matrix shape:", X_scaled.shape)
```

```python
from sklearn.cluster import DBSCAN

# Create DBSCAN model
dbscan = DBSCAN(
    eps=0.5,
    min_samples=10,
    n_jobs=-1
)

# Fit DBSCAN
dbscan_labels = dbscan.fit_predict(X_scaled)

print("DBSCAN completed!")
print("Total data points:", len(dbscan_labels))
print("Clusters found:", len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0))
print("Noise/anomalies:", np.sum(dbscan_labels == -1))
```
